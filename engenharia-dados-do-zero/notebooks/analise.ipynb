```python
import sqlite3
```

```python
conn = sqlite3.connect('../data/vendas.db')
```

```python
df.to_sql(
    "vendas",
    conn,
    if_exists="replace",
    index=False
)
```

```python
query = """
SELECT *
FROM vendas
LIMIT 5
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT
    cidade,
    SUM(valor_total) AS faturamento
FROM vendas
GROUP BY cidade
ORDER BY faturamento DESC
"""

pd.read_sql(query, conn)
```
